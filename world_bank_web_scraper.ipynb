# World Bank Country Data Web Scraper

A Python web scraper that extracts country-level economic and demographic data from the World Bank website using Requests and BeautifulSoup.

## Data Collected

- Country
- Life Expectancy
- Poverty
- Population
- Population Growth
- Net Migration
- GDP

```python
# import all libraries
import requests
from bs4 import BeautifulSoup
# for url
from urllib.parse import urljoin
import csv
```

# 1. Extract Country Links

This function collects the URLs of individual countries from the World Bank country directory.

```python
# create function to get the links of all countries
def get_links(url):
  # create list for all urls/ links
  link_list=[]
  # base url
  first_part_of_link='https://data.worldbank.org'
  response=requests.get(url)
  soup= BeautifulSoup(response.text,"html.parser")
  # find all alphabet sections
  sections=soup.find_all('section',class_='nav-item')
  for section in sections:
    # each country in each each section
    li=section.find_all('li')
    for i in li:
      # country href
      link=(i.find('a')['href'])
      # combine base url and href
      complete_link=urljoin(first_part_of_link,link)
      # store it in list
      link_list.append(complete_link)
      # return list
  return link_list
```

## 2. Extract Country Data

- Net migration

The extracted country data is stored in a CSV file for further analysis and processing.
